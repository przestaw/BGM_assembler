```python
genomes = [[i, j] for i, j in files.load_fasta_file('../fasta/E.Coli/GCF_000005845.2_ASM584v2_genomic.fna')]

genomes[0][1] = dig.digest_with_enzyme(genomes[0][1], enzyme='BspQI')
```

```python
import src.generator_util as gen

e_coli_map = genomes

def gen_test(quant = 1000):
    e_coli_map[0][1] = dig.gen_bin_map_from_digest(genomes[0][1], quant)

    for i in range(200, 500, 50):
        example = gen.generate_errorfree_overlaping(e_coli_map[0], i, 100)
        files.write_silico_maps(example, './perfect/ecoli_'+str(quant)+'quant_'+str(i)+'len_100shift_errorfree.binmap')

    for i in range(100, 350, 50):
        example = gen.generate_errorfree_overlaping(e_coli_map[0], i, 50)
        files.write_silico_maps(example, './perfect/ecoli_'+str(quant)+'quant_'+str(i)+'len_50shift_errorfree.binmap')

    for i in range(200, 500, 50):
        example = gen.generate_errorfree_distribued(e_coli_map[0], i, 75)
        files.write_silico_maps(example, './normal/ecoli_'+str(quant)+'quant_normal_'+str(i)+'len_75var_errorfree.binmap')

    for i in range(120, 400, 50):
        example = gen.generate_errorfree_distribued(e_coli_map[0], i, 45)
        files.write_silico_maps(example, './normal/ecoli_'+str(quant)+'quant_normal_'+str(i)+'len_45var_errorfree.binmap')

gen_test(500)
gen_test(1000)
gen_test(2000)
```
